# Khan Hotel - Restaurant Billing System

A menu-driven ordering program that takes items (Half/Full/Per-piece), adds 5% GST and prints a formatted bill.

**How to use:** run the cells from top to bottom. In the ordering cell, enter a serial number and quantity for each item, and type `-1` when you are done.

## 1. Menu data and GST rate

```python
GST_RATE = 0.05  # 5% GST

# serial no. -> (name, half price, full price, per-piece price)
# half is None when no half portion exists; piece is not None for per-piece items
MENU = {
    1: ("Butter Chicken", 120, 230, None),
    2: ("Malai Kofta", None, 130, None),
    3: ("Nihari", 200, 400, None),
    4: ("Chicken Malai Tikka", 80, 140, None),
    5: ("Afghani Chicken", 90, 150, None),
    6: ("Hyderabadi Biryani", 70, 140, None),
    7: ("Moradabadi Biryani", 70, 140, None),
    8: ("Seekh Kabab", None, None, 30),
    9: ("Plain Naan", None, None, 20),
    10: ("Butter Naan", None, None, 40),
}
```

## 2. Display the menu

```python
def display_menu():

    print("\t          ****KHAN HOTEL****\n")
    print("\t\t       **MENU**\n")
    print("Sr. No. \t  Food Name \t\t   Price")
    print("\t\t\t\t\tHalf\tFull")
    print("1. \t\tButter Chicken \t\t\u20B9120\t\u20B9230")
    print("2. \t\tMalai Kofta \t\t -\t\u20B9130")
    print("3. \t\tNihari \t\t\t\u20B9200\t\u20B9400")
    print("4. \t\tChicken Malai Tikka \t\u20B980\t\u20B9140")
    print("5. \t\tAfghani Chicken \t\u20B990\t\u20B9150")
    print("6. \t\tHyderabadi Biryani \t\u20B970\t\u20B9140")
    print("7. \t\tMoradabadi Biryani \t\u20B970\t\u20B9140")
    print("8. \t\tSeekh Kabab \t\t\u20B930(Per Piece)")
    print("9. \t\tPlain Naan \t\t\u20B920(Per Piece)")
    print("10. \t\tButter Naan \t\t\u20B940(Per Piece)")
```

## 3. Helper functions

- `get_portion` asks for Half/Full where both exist and returns the portion label and unit price.
- `calculate_bill` computes line amounts, subtotal, GST and total.
- `generate_bill` prints the formatted bill.

```python
def get_portion(n):
    """Ask Half/Full where both exist; return (label, unit_price)."""
    name, half, full, piece = MENU[n]

    if piece is not None:
        return "Piece", piece
    if half is None:
        return "Full", full

    while True:
        choice = input("Half or Full portion? (H/F): ").strip().upper()
        if choice == "H":
            return "Half", half
        if choice == "F":
            return "Full", full
        print("Please enter H for Half or F for Full.")
```

```python
def calculate_bill(order):
    """order: list of (serial_no, portion, unit_price, quantity).
    Returns (line_items, subtotal, gst, total)."""
    lines = []
    subtotal = 0

    for n, portion, price, qty in order:
        amount = price * qty
        subtotal += amount
        lines.append((MENU[n][0], portion, price, qty, amount))

    gst = round(subtotal * GST_RATE, 2)
    total = round(subtotal + gst, 2)
    return lines, subtotal, gst, total
```

```python
def generate_bill(lines, subtotal, gst, total):
    print("\n" + "=" * 62)
    print("\t\t       ****KHAN HOTEL****")
    print("\t\t\t    BILL")
    print("=" * 62)
    print(f"{'Item':<24}{'Type':<8}{'Rate':>8}{'Qty':>6}{'Amount':>12}")
    print("-" * 62)

    for name, portion, price, qty, amount in lines:
        print(f"{name:<24}{portion:<8}{price:>8}{qty:>6}{amount:>12.2f}")

    print("-" * 62)
    print(f"{'Subtotal':<46}\u20B9{subtotal:>13.2f}")
    print(f"{'GST (5%)':<46}\u20B9{gst:>13.2f}")
    print("=" * 62)
    print(f"{'TOTAL':<46}\u20B9{total:>13.2f}")
    print("=" * 62)
    print("\tThank you! Please visit again.\n")
```

## 4. Take the order

Run this cell, then enter the serial number and quantity for each item. Enter `-1` to finish.

```python
display_menu()

print("\nINSTRUCTION: WRITE -1 WHEN YOU FINISH GIVING ORDER")
print("\nPlease give your order")

order = []

n = 0

while n != -1:
    try:
        n = int(input("\nPlease write the serial number of item you want to order: "))
    except ValueError:
        print("Please enter a number.")
        continue

    if 1 <= n <= 10:
        try:
            quantity = int(input("Please enter the quantity you want to order: "))
        except ValueError:
            print("Please enter a valid quantity.")
            continue

        if quantity <= 0:
            print("Quantity must be at least 1.")
            continue

        portion, price = get_portion(n)
        order.append((n, portion, price, quantity))
        print(f"Added: {quantity} x {MENU[n][0]} ({portion})")

    elif n == -1:
        print("Thanks for ordering from our restaurant.")
        print("Hope you enjoy the food.")
        print("Your bill will be generated shortly.")

    else:
        print("Please select the valid number.")
```

## 5. Generate the bill

```python
if order:
    lines, subtotal, gst, total = calculate_bill(order)
    generate_bill(lines, subtotal, gst, total)
else:
    print("No items ordered. No bill generated.")
```
